```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

df = pd.read_csv(filepath_or_buffer="./data1.csv")
#print(data_frame.info(), "\n")
fig, axes = plt.subplots(1, 3, figsize = (15, 5))
axes[0].scatter(df['x1'], df['y'])
axes[0].set_title('x1 vs y')
axes[0].set_xlabel('x1')
axes[0].set_ylabel('y')

axes[1].scatter(df['x2'], df['y'])
axes[1].set_title('x2 vs y')
axes[1].set_xlabel('x2')

axes[2].scatter(df['x1'], df['x2'])
axes[2].set_title('x1 vs x2')
axes[2].set_xlabel('x1')
axes[2].set_ylabel('x2')
```

jak widzimy w 1 wykresie jak x rosnie to y rosnie o mniej wiecej tyle samo stad możemy dojść do wniosku ze 
y ≈ x1 natomiast z drugiego wykresu możemy w ten sam sposo” dojsc do wniosku ze y ≈ -x2 stąd y = x1-x2 zwizualizujmy to:

```python
plt.scatter(df['x1'] - df['x2'], df['y'])
plt.show()
```

```python
def preproces_data(X_train, y_train):
    diff = abs(X_train['x1'] - X_train['x2'] -y_train)
    mask_good = diff < 500
    X_train_good = X_train[mask_good]
    y_train_good = y_train[mask_good]
    return X_train_good, y_train_good


dataX, dataY = preproces_data(df[['x1', 'x2']], df['y'])
plt.scatter(dataX['x1'] - dataX['x2'], dataY)
plt.show()
print(f"before records: {len(df)}, after: {len(dataX)}")
```
