## 2.2
Model baselinowy to będzie po rpsotu średnia z y na zbiorze treningowym jeżeli nasz modle będzie gorszy od niej to jest troche bezurzyteczny

```python
import pandas as pd
import matplotlib.pyplot as plt
import sklearn.model_selection as sk
from sklearn.linear_model import LinearRegression
from sklearn.dummy import DummyRegressor
from sklearn.metrics import mean_squared_error

df = pd.read_csv(filepath_or_buffer="./data1.csv")

def preproces_data(X_train, y_train):
    diff = abs(X_train['x1'] - X_train['x2'] -y_train)
    mask_good = diff < 500
    X_train_good = X_train[mask_good]
    y_train_good = y_train[mask_good]
    return X_train_good, y_train_good
X_train, X_test, y_train, y_test = sk.train_test_split(df[['x1', 'x2']], df['y'], test_size=0.2, random_state=2137)
X_train_pp,y_train_pp = preproces_data(X_train, y_train)

model_LR = LinearRegression()
model_LR_base = DummyRegressor(strategy='mean')
model_LR_pp =  LinearRegression()
model_LR_pp_base = DummyRegressor(strategy='mean')

model_LR.fit(X_train, y_train)
model_LR_base.fit(X_train, y_train)
model_LR_pp.fit(X_train_pp, y_train_pp)
model_LR_pp_base.fit(X_train_pp, y_train_pp)
```

```python
pred_LR = model_LR.predict(X_test)
pred_LR_base = model_LR_base.predict(X_test)
pred_LR_pp = model_LR_pp.predict(X_test)
pred_LR_pp_base = model_LR_pp_base.predict(X_test)

mse_LR = mean_squared_error(y_test, pred_LR)
mse_LR_base = mean_squared_error(y_test, pred_LR_base)
mse_LR_pp = mean_squared_error(y_test, pred_LR_pp)
mse_LR_pp_base = mean_squared_error(y_test, pred_LR_pp_base)

wyniki = {
    "Model": ["Baseline (Średnia)", "Regresja Liniowa"],
    "MSE (Surowe dane)": [mse_LR_base, mse_LR],
    "MSE (Dane po preprocessingu)": [mse_LR_pp_base, mse_LR_pp]
}

df_wyniki = pd.DataFrame(wyniki)

df_wyniki
```
